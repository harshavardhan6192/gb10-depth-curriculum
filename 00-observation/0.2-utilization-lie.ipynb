```python
import torch, time

def busy_compute(n=8192, iters=1000):
    a = torch.randn(n, n, device='cuda', dtype=torch.bfloat16)
    b = torch.randn(n, n, device='cuda', dtype=torch.bfloat16)
    torch.cuda.synchronize()
    t0 = time.perf_counter()
    for _ in range(iters):
        c = a @ b
    torch.cuda.synchronize()
    print(f"A (compute): {time.perf_counter()-t0:.2f}s")

def busy_memory(n=256_000_000, iters=1000):
    a = torch.empty(n, device='cuda', dtype=torch.bfloat16)
    b = torch.empty(n, device='cuda', dtype=torch.bfloat16)
    torch.cuda.synchronize()
    t0 = time.perf_counter()
    for _ in range(iters):
        b.copy_(a)
    torch.cuda.synchronize()
    print(f"B (memory): {time.perf_counter()-t0:.2f}s")

def busy_launch(iters=2_000_000):
    a = torch.ones(32, device='cuda')
    torch.cuda.synchronize()
    t0 = time.perf_counter()
    for _ in range(iters):
        a.add_(1.0)
    torch.cuda.synchronize()
    print(f"C (launch): {time.perf_counter()-t0:.2f}s")
```
